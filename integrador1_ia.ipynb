```python
TIPOS = ("notebook", "tablet", "kit arduino", "fone")

equipamentos = [
    {"codigo": "NB01", "tipo": "notebook", "disponivel": True,
     "com": "", "emprestimos": 4},
    {"codigo": "NB02", "tipo": "notebook", "disponivel": False,
     "com": "Ana", "emprestimos": 7},
    {"codigo": "TB01", "tipo": "tablet", "disponivel": True,
     "com": "", "emprestimos": 2},
    {"codigo": "AR01", "tipo": "kit arduino", "disponivel": False,
     "com": "Bruno", "emprestimos": 5},
    {"codigo": "FN01", "tipo": "fone", "disponivel": True,
     "com": "", "emprestimos": 0},
]
```

```python
def buscar(codigo):
    #Retorna o equipamento com o código informado ou lança um erro.
    codigo = codigo.strip().upper()

    for equipamento in equipamentos:
        if equipamento["codigo"] == codigo:
            return equipamento

    raise ValueError(f"Não existe equipamento com o código {codigo!r}.")
```

```python
def listar():
    #Mostra todos os equipamentos cadastrados.
    if not equipamentos:
        print("Não há equipamentos cadastrados.")
        return

    print("\nEquipamentos do laboratório")
    for equipamento in equipamentos:
        if equipamento["disponivel"]:
            situacao = "Disponível"
        else:
            situacao = f"Emprestado para {equipamento['com']}"

        print(
            f"Código: {equipamento['codigo']} | "
            f"Tipo: {equipamento['tipo']} | "
            f"Situação: {situacao} | "
            f"Empréstimos: {equipamento['emprestimos']}"
        )

```

```python
def cadastrar():
    #Cadastra um equipamento novo, disponível e com zero empréstimos.
    tipo = input(f"Tipo ({', '.join(TIPOS)}): ").strip().lower()
    if tipo not in TIPOS:
        raise ValueError("Tipo inválido. Escolha um dos tipos permitidos.")

    codigo = input("Código novo: ").strip().upper()
    if not codigo:
        raise ValueError("O código não pode ficar em branco.")

    for equipamento in equipamentos:
        if equipamento["codigo"] == codigo:
            raise ValueError(f"O código {codigo} já está cadastrado.")

    novo_equipamento = {
        "codigo": codigo,
        "tipo": tipo,
        "disponivel": True,
        "com": "",
        "emprestimos": 0,
    }
    equipamentos.append(novo_equipamento)
    print(f"Equipamento {codigo} cadastrado com sucesso!")
```

```python
def verificar(nome):
    #Impede que um aluno tenha mais de dois equipamentos emprestados.
    nome_normalizado = nome.strip().casefold()
    quantidade = sum(1 for equipamento in equipamentos if not equipamento["disponivel"]and equipamento["com"].strip().casefold() == nome_normalizado)

    if quantidade >= 2:
        raise ValueError(
            f"{nome} já está com {quantidade} equipamentos emprestados. "
            "Cada aluno pode ficar com no máximo dois."
        )
```

```python
def emprestar():
    #Registra o empréstimo de um equipamento disponível.
    codigo = input("Código do equipamento: ").strip().upper()
    equipamento = buscar(codigo)

    if not equipamento["disponivel"]:
        raise ValueError(
            f"O equipamento {codigo} já está emprestado para {equipamento['com']}."
        )

    nome = input("Nome do aluno: ").strip()
    if not nome:
        raise ValueError("O nome do aluno não pode ficar em branco.")

    verificar(nome)

    equipamento["disponivel"] = False
    equipamento["com"] = nome
    equipamento["emprestimos"] += 1
    print(f"Empréstimo do equipamento {codigo} registrado para {nome}.")

```

```python
def devolver():
    #Registra a devolução de um equipamento emprestado.
    codigo = input("Código do equipamento: ").strip().upper()
    equipamento = buscar(codigo)

    if equipamento["disponivel"]:
        raise ValueError(f"O equipamento {codigo} já está disponível; não está emprestado.")

    nome = equipamento["com"]
    equipamento["disponivel"] = True
    equipamento["com"] = ""
    print(f"Equipamento {codigo} devolvido por {nome}.")
```

```python
def consultar():
    #Mostra todos os dados de um equipamento a partir do código.
    codigo = input("Código do equipamento: ").strip().upper()
    equipamento = buscar(codigo)

    print("\n--- Dados do equipamento ---")
    for chave, valor in equipamento.items():
        print(f"{chave.capitalize()}: {valor}")
```

```python
def relatorio():
    #Mostra totais, disponibilidade, empréstimos e quantidades por tipo.
    total = len(equipamentos)
    disponiveis = sum(1 for equipamento in equipamentos if equipamento["disponivel"])
    emprestados = total - disponiveis

    print("\n Relatório do laboratório")
    print(f"Total de equipamentos: {total}")
    print(f"Equipamentos disponíveis: {disponiveis}")
    print(f"Equipamentos emprestados: {emprestados}")

    if equipamentos:
        maior_quantidade = max(equipamento["emprestimos"] for equipamento in equipamentos)
        mais_emprestados = [
            equipamento for equipamento in equipamentos
            if equipamento["emprestimos"] == maior_quantidade
        ]
        print("Equipamento(s) mais emprestado(s):")
        for equipamento in mais_emprestados:
            print(
                f"- {equipamento['codigo']} ({equipamento['tipo']}): "
                f"{equipamento['emprestimos']} empréstimos"
            )
    else:
        print("Ainda não há equipamentos para comparar.")

    print("Quantidade por tipo:")
    for tipo in TIPOS:
        quantidade = sum(1 for equipamento in equipamentos if equipamento["tipo"] == tipo)
        print(f"- {tipo}: {quantidade}")
```

```python
def excluir():
    #Remove um equipamento, desde que ele não esteja emprestado.
    codigo = input("Código do equipamento: ").strip().upper()
    equipamento = buscar(codigo)

    if not equipamento["disponivel"]:
        raise ValueError(
            f"Não é possível remover {codigo}: ele está emprestado para {equipamento['com']}."
        )

    confirmar = input(
        f"Gostaria de excluir o equipamento {equipamento['codigo']}? (sim/não): "
    ).strip().lower()

    if confirmar in ("sim", "s"):
        equipamentos.remove(equipamento)
        print(f"Equipamento {codigo} removido com sucesso.")
    else:
        print("Exclusão cancelada.")
```

```python

def sair():
    #Encerra o programa.
    print("Saindo do programa...")
```

```python
def menu():
    #Exibe o menu e executa as opções até o usuário escolher sair.
    opcoes = {
        "1": listar,
        "2": cadastrar,
        "3": emprestar,
        "4": devolver,
        "5": consultar,
        "6": relatorio,
        "7": excluir,
    }

    while True:
        print(
            "\n Controle do Sistema \n"
            "1 - Listar equipamentos\n"
            "2 - Cadastrar equipamento\n"
            "3 - Emprestar equipamento\n"
            "4 - Devolver equipamento\n"
            "5 - Consultar equipamento\n"
            "6 - Relatório\n"
            "7 - Remover equipamento\n"
            "0 - Sair"
        )

        opcao = input("Escolha uma opção: ").strip()

        if opcao == "0":
            sair()
            break

        if opcao not in opcoes:
            try:
                raise ValueError("Opção inválida. Digite um número de 0 a 7.")
            except ValueError as erro:
                print(f"Erro: {erro}")
            continue

        try:
            opcoes[opcao]()
        except ValueError as erro:
            print(f"Erro: {erro}")
        except (EOFError, KeyboardInterrupt):
            print("\nEntrada interrompida. Voltando ao menu.")

```

```python
menu()
```
